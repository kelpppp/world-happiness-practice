Rename column headers to all lowercase and without spaces, so they can be more conveniently accessed later.

```python
mapping = {
    'Overall rank' : 'overall_rank',
    'Country or region' : 'country',
    'Score' : 'score',
    'GDP per capita' : 'gdp_per_capita',
    'Social support' : 'social_support',
    'Healthy life expectancy' : 'life_expectancy',
    'Freedom to make life choices' : 'freedom',
    'Generosity' : 'generosity',
    'Perceptions of corruption' : 'perceived_corruption'
}

df_2019 = df_2019.rename(columns=mapping)

df_2019.head()
```

## 5. Analysis: Sorting & Filtering

**Q: Create a dataframe of the 10 happiest countries, keeping only the rank, country name, and score columns**

```python
# get top 10 ranked countries using boolean indexing
df_top_10_countries = df_2019[df_2019['overall_rank'] <= 10]

# alternatively: df_2019.head(10)

# keep only desired columns by indexing with list of column names
df_top_10_countries = df_top_10_countries[['overall_rank', 'country', 'score']]

# reset index to start from 0, and drop old index column
df_top_10_countries.reset_index(drop=True, inplace=True)

df_top_10_countries
```

**Q: Create a dataframe of the 10 least happy countries.**

```python
# get lowest 10 ranked countries
df_lowest_10_countries = df_2019.tail(10)

# keep only desired columns by indexing with list of column names
df_lowest_10_countries = df_lowest_10_countries[['overall_rank', 'country', 'score']]

# reset index to start from 0, drop old index colum
df_lowest_10_countries.reset_index(drop=True, inplace=True)

df_lowest_10_countries
```

**Q: Among countries with GDP per capita above the median, which 10 have the lowest happiness score?**

```python
# calculate median GDP per capita
median_gdp = df_2019['gdp_per_capita'].median()

# get countries with GDP per capita above the median
df_gdp_above_median = df_2019[df_2019['gdp_per_capita'] > median_gdp]

# among those, get 10 countries with lowest happiness score
df_gdp_above_median.nsmallest(10, 'score')

# alternatively: df_gdp_above_median.tail(10)
```

`nsmallest()` returns the `n` rows with the smallest values from a dataframe or series
* Syntax: `df.nsmallest(n, columns, keep)`
* Parameters: 
    * `n`: number of items to retrieve
    * `columns`: column name, or list of column names
    * `keep`: specifies what to do with duplicate rows
