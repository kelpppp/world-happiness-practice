## 6. Analysis: Grouping

**Q: Split countries into low, medium, and high "tiers" based on the GDP per capita score contribution (hint: using pd.cut or pd.qcut to create the buckets). What's the average happiness score in each tier?**

```python
df_2019['gdp_per_capita'].describe()
```

```python
# create labels for tier categories
labels = ['Low', 'Medium', 'High']

# get a subset of df_2019 containing this list of columns
df_countries_by_gdp = df_2019[['overall_rank', 'country', 'score', 'gdp_per_capita']]

# divide gdp values into 3 equally sized tiers (low, medium, high)
# add a new column for these tiers
df_countries_by_gdp['gdp_tier'] = pd.qcut(df_2019['gdp_per_capita'], q=3, labels=labels)

df_countries_by_gdp
```

```python
# group the countries by the tier values
# for each group (low, medium, high), calculate the mean happiness score
df_countries_by_gdp.groupby('gdp_tier')['score'].mean() # returns a Series
```

```python
# convert series returned from groupby and aggregation into a dataframe
df_avg_score_by_gdp = df_countries_by_gdp.groupby('gdp_tier')['score'].mean().reset_index()
df_avg_score_by_gdp
```

* Countries were split into 3 equally-sized tiers based on GDP per capita. This was done using the `pd.qcut()` function, which splits a Series into approximately equally sized bins.
* A dataframe with only the `overall_rank`, `country`, `score`, and `gdp_per_capita` columns was obtained. A new `gdp_tier` column was added to display the GDP tier of each country.
* The countries were grouped by their `gdp_tier` column values. For each tier group (low, medium, high), the average happiness score was calculated by applying the `.mean()` function on the `score` column.
* The Series returned from the `groupby()` and `mean()` operations was converted to a DataFrame by applying the `.reset_index()` function. The index labels (0, 1, 2...) were pulled into a new column, changing the data structure from 1D (Series) to 2D (DataFrame).

Result: 
* Mean score for low tier countries: 4.43
* Mean score for medium tier countries: 5.32
* Mean score for high tier countries: 6.47

**Q: Repeat the same steps for the other 5 factors. Store the results in their own DataFrames.**

```python
labels = ['Low', 'Medium', 'High']
factors = ['gdp_per_capita', 'social_support', 'life_expectancy', 'freedom', 'generosity', 'perceived_corruption']
# created dictionary of dataframes
dfs_avg_score_by_tier = {}

# for each factor, dynamically create its dataframe of average score by tier
# store each created dataframe into a dictionary of dataframes
for factor in factors:
    # create dataframe
    dfs_avg_score_by_tier[f'df_avg_score_by_{factor}'] = df_2019.groupby(
        pd.qcut(df_2019[factor], q=3, labels=labels)
        )['score'].mean().reset_index()

# verify by accessing a dataframe in dataframes dictionary
dfs_avg_score_by_tier['df_avg_score_by_social_support'].head()
```

* The goal was to create 6 DataFrames that show the average score by tier, one for each factor. To do so, I looped through each factor in a list of factors, and inserted the factor name where needed dynamically using f-strings. 
* The line that creates the DataFrame is complex-looking, but it's just the combined version of what was done previously for the GDP per capita, but with the added use of f-strings. 

*A note on the six factors:*

The six factors are variables used *after* obtaining the survey results to *explain* the variation in happiness scores between countries. The numerical score assigned to each factor shows how much that specific condition contributes - positively or negatively - to a country's happiness score, compared to the hypothetical baseline Dystopia. 

**Q: For each of the six factors, how big is the jump between the "Low" tier and "High" tier average happiness score? This might reveal which factor most distinguishes happy from unhappy countries.**

```python
factors = ['gdp_per_capita', 'social_support', 'life_expectancy', 'freedom', 'generosity', 'perceived_corruption']

# initialize list of lists, containing rows, to be converted to dataframe
factor_gaps_list = []

# for each of the 6 factors, get the score gap between the high and low tier average scores
for factor in factors:
    df = dfs_avg_score_by_tier[f'df_avg_score_by_{factor}']
    high = df.iat[2, 1]
    low = df.iat[0, 1]
    # add this score gap data for this factor as a list, to be converted into a row
    factor_gaps_list.append([factor, high - low])

# convert list of lists, containing rows, into a dataframe; specify column names
# the dataframe displays, for each factor, the gap in happiness scores between high and low tiers
df_score_gaps = pd.DataFrame(factor_gaps_list, columns=['factor', 'score_gap']) \
                            .sort_values('score_gap', ascending=False) \
                            .reset_index(drop=True)

df_score_gaps
```

The goal was to create a DataFrame displaying, for each factor, the gap in average happiness scores between the high and low country tiers.
* A list of lists, `factor_gap_list`, was initialized to store the data - the rows - that would be converted to the DataFrame. Each row contains the factor and the score gap for that factor.
* Iterating through each of the 6 factors, the `df_average_score_by_{factor}` DataFrame for that factor was accessed from the list of DataFrames (created in the previous segment).
    * The score gap was calculated by subtracting the score of the high tier from that of the low tier.
    * The scores for each the high and low tiers were accessed using the `.iat[rowIndex, colIndex]` function, which is used to access or modify a single value by its index position. 
    * This row  - th factor and the score gap - was appended to the list of lists data variable.
* The data gathered was converted into a DataFrame. To do this, the `factor_gaps_list` list of lists was passed as the `data` parameter into the `pd.Dataframe(data)` function. A list of column names were passed into the `columns` parameter, because that information is not given in the `data` list of lists. 

Result:
* Life expectancy was the factor with the greatest score gap: 2.153
* Social support: 2.094
* GDP per capita: 2.046
* Freedom: 1.355
* Perceived corruption: 0.604
* Generosity was the factor with the lowest score gap: 0.246

**Q: For each country, find which of the six factor columns has the largest value — creating a new categorical column like "dominant factor." Then groupby that and count how many countries are "led" by each factor. Are there a few factors that most countries are mainly driven by, or is it more mixed?**

```python
# for each country, find the maximum value between the factor columns
factors = ['gdp_per_capita', 'social_support', 'life_expectancy', 'freedom', 'generosity', 'perceived_corruption']
df_2019['dominant_factor'] = df_2019[factors].idxmax(axis=1)

df_2019
```

The `idxmax()` method returns, by default, a Series with the index of the maximum value for each column. 
* By default, the method searches down rows (row-wise) to find index of the maximum value in a *column*. 
* By specifying `axis=1` (or `axis=columns`), the method searches across columns (column-wise) to find index of the maximum value in a  *row*.

```python
# group countries by dominant factor
# count the number of countries led by each dominant factor
df_countries_per_dominant_factor = df_2019['dominant_factor'].value_counts().reset_index()

# add missing rows of dominant factors (with 0 number of countries)
missing_data = [
    {'dominant_factor': 'life_expectancy', 'count' : 0},
    {'dominant_factor': 'freedom', 'count' : 0},
    {'dominant_factor': 'perceived_corruption', 'count' : 0},
]
df_missing_data = pd.DataFrame(missing_data)

# concatenate missing rows dataframe with dominant factors dataframe
df_countries_per_dominant_factor = pd.concat([df_countries_per_dominant_factor, df_missing_data], ignore_index=True) \
                                     .sort_values('count', ascending=False) \
                                     .reset_index(drop=True)

# rename dominant_factor column values to be more readable
mapping = {'social_support' : 'Social support', 
           'gdp_per_capita' : 'GDP per capita',
           'generosity' : 'Generosity', 
           'life_expectancy' : 'Life expectancy',
           'freedom' : 'Freedom', 
           'perceived_corruption' : 'Perceived corruption'}
df_countries_per_dominant_factor['dominant_factor'] = df_countries_per_dominant_factor['dominant_factor'].replace(mapping)

df_countries_per_dominant_factor
```

* The 'dominant factor' is the factor with the highest score of the six for a country - the factor that contributed the most to increasing its happiness score. 
* The 'dominant factor' for each country was found by applying the `idxmax()` function, which returns a Series with the index of the maximum value for each row. This Series was used to create a new column, `dominant_factor`. 
* The number of countries for each dominant factor was counted, and this count was used to create a new DataFrame showing the number of countries in each `dominant_factor` group. 
    * This was done by applying the `value_counts()` function to count the occurences of each unique value in the `dominant_factors` column.
    * The `.reset_index()` function was used to return a new DataFrame.

* The `df_countries_per_dominant_factor` DataFrame did not include all factors, because some of them were not dominant factors for any of the countries. However, for completeness, I added the missing factors (with a `count` of 0) to the DataFrame.
    * This was done by creating a list of dictionaries for the rows I wanted to add, which was transformed into a Pandas DataFrame. Then, using `pd.concat()`, the DataFrame created for the new rows was merged with the `df_countries_per_dominant_factor` DataFrame from earlier.
    * The `dominant_factor` column values were renamed for readability. This was done by passing a dictionary mapping the old names to the new names into the `replace()` function, which was applied to the `dominant_factor` column of the DataFrame.

**Q: Merge two years' dataframes on the country name column. How many countries failed to match?**

```python
# read in 2018 data and store in dataframe
df_2018 = pd.read_csv('data/2018.csv')

# for consistency, rename columns the same way as for df_2019
mapping = {
    'Overall rank' : 'overall_rank',
    'Country or region' : 'country',
    'Score' : 'score',
    'GDP per capita' : 'gdp_per_capita',
    'Social support' : 'social_support',
    'Healthy life expectancy' : 'life_expectancy',
    'Freedom to make life choices' : 'freedom',
    'Generosity' : 'generosity',
    'Perceptions of corruption' : 'perceived_corruption'
}

df_2018 = df_2018.rename(columns=mapping)

df_2018
```

```python
# first drop 'dominant_factor' column, as it was not calculated for the 2018 data
df_2019_to_merge = df_2019.drop(columns='dominant_factor')

# merge the two dataframes on the 'country' column
# add the suffix '_2018' or '_2019' to columns with the same names in both dataframes
df_merged = pd.merge(df_2018, df_2019_to_merge, how='inner', on='country', suffixes=['_2018', '_2019'])

# this returns a list (new_columns) that moves the 'country' column to the front
new_columns = ['country'] + [col for col in df_merged.columns if col != 'country']
df_merged = df_merged[new_columns]

df_merged.head()
```

The `pd.merge()` function is used to combine two DataFrames based on common columns of indices. 

Syntax: `pd.merge(left, right, how, on, left_on, right_on, left_index, right_index, suffixes=())`

Parameters:
* `left`, `right`: The two DataFrames to be joined.
* `how` (default `inner`): The type of join to perform. The default `inner` keeps only the rows where the matching key exists in Dataframes.
* `on`: The column name(s) to join on, which must exist in both DataFrames. 
* `left_on`, `right_on`: Used if the column(s) to join on have different names in the left and right DataFrames.
* `left_index`, `right_index`: Set to `True` if want to join using row index instead of columns.
* `suffixes`: A tuple specifying what to append to overlapping column names.

```python
# check for whether there are any missing values for any scores

# return a series (mask) of true or false values indicating whether any rows have NaN values
mask = df_merged.isna().any(axis=1)
# show dataframe with only the rows that have NaN values
df_merged[mask]
```

* The goal was to merge the 2018 and 2019 datasets, and to clean up the merged dataset for readability.
* The `dominant_factor` column previously created was dropped from the 2019 dataset, as they were not calculated for the 2018 dataset. 
    * Used the `drop()` function, specifying the column name in the `columns` parameter.
* The 2018 and 2019 DataFrames were merged using `pd.merge()`.
    * The DataFrames were joined on the `country` column, and the suffixes `_2018` and `_2019` were added to overlapping column names. 
* Originally, the merged DataFrame had the `overall_ranking_2019` column at the front. It made more sense to have `country` as the first column instead.
    * List comprehension was used to bring the `country` column to the front. A `new_columns` list was created to store the reordered columns, which was passed as an index into the `df_merged` DataFrame, returning a DataFrame with columns in the specified order. 
* The `isna()` and `any()` functions were chained and applied to the `country` column to check for missing countries. 
* The `isna()` and `any()` functions were chained and applied to the `df_merged` DataFrame, specifying `axis=1`. This returned a Series of `True` or `False` values indicating the presence of `NaN` values in any of the rows. This Series was used as a mask for boolean indexing to get only the rows that have `NaN` values.

**Q: For countries present in both years, which had the biggest increase in total happiness score? Biggest decrease? For the countries with the biggest score increase / decrease, which specific factors changed the most?**

```python
# create a new column showing the increase / decrease in scores from 2018 to 2019
df_merged['score_increase'] = df_merged['score_2019'] - df_merged['score_2018']
df_merged[['country', 'score_increase']].sort_values(by='score_increase', ascending=False).reset_index(drop=True)
```

```python
# get country row with greatest score increase
max_diff_row = df_merged[df_merged['score_increase'] == df_merged['score_increase'].max()]
display(max_diff_row)

# get country row with greatest score decrease
min_diff_row = df_merged[df_merged['score_increase'] == df_merged['score_increase'].min()]

display(min_diff_row)
```

```python
factors = ['gdp_per_capita', 'social_support', 'life_expectancy', 'freedom', 'generosity', 'perceived_corruption']

# make a list of column names for each factor increase score column (ex: 'freedom_increase')
factor_increases_names = []
for factor in factors:
    factor_increases_names.append(f'{factor}_increase')
```

```python
# Burundi:

# create dataframe containing only columns for factor increase scores
df_burundi_factor_increases = max_diff_row[factor_increases_names]
# get list of factor increase scores from created dataframe
factor_increases_values = df_burundi_factor_increases.iloc[0].tolist()

# create dict of lists to store data to be converted into dataframe
    # pass list of factor names as values in 'factor' column
    # pass list of factor increase scores as values in 'score_increase' column
burundi_data = {
    "factor": factors,
    "score_increase": factor_increases_values
}

# create new dataframe for burundi with the dict
df_burundi_factors_diff = pd.DataFrame(burundi_data) \
                            .sort_values(by='score_increase', ascending=False).reset_index(drop=True)
df_burundi_factors_diff
```

```python
# Malaysia:

# create dataframe containing only columns for factor increase scores
df_malaysia_factor_increases = min_diff_row[factor_increases_names]
# get list of factor score increases from created dataframe
factor_increases_values = df_malaysia_factor_increases.iloc[0].tolist()

# create dict of lists to store data to be converted into dataframe
    # pass list of factor names as values in 'factor' column
    # pass list of factor increase scores as values in 'score_increase' column
malaysia_data = {
    "factor": factors,
    "score_increase": factor_increases_values
}

# create new dataframe for malaysia with the dict
df_malaysia_factors_diff = pd.DataFrame(malaysia_data) \
                             .sort_values(by='score_increase', ascending=False).reset_index(drop=True)
df_malaysia_factors_diff
```

Since there is only one row of values in the `max_diff_row` or `min_diff_row` DataFrames, `iloc[0]` was applied to the DataFrame subset to access the first row (index 0), and `tolist()` was chained to get those rows values as a list.

***Technique improvement:***

**What I did:** In the code, I **transposed** (flip the rows and columns) of Dataframes: I converted the two single-row DataFrames for Burundi and Malaysia to two 6 row DataFrames containing the score increases for each factor. I did this by extracting the values from the single-row DataFrames, and constructing the new DataFrame from a dictionary of lists. There is a simpler way to do this, becuase **transposing** DataFrames is a very common practice. 

```python
factors = ['gdp_per_capita', 'social_support', 'life_expectancy', 'freedom', 'generosity', 'perceived_corruption']
df_burundi_factor_increases = max_diff_row[factor_increases_names]
factor_increases_values = df_burundi_factor_increases.iloc[0].tolist()

burundi_data = {
    "factor": factors,
    "score_increase": factor_increases_values
}

df_burundi_factors_diff = pd.DataFrame(burundi_data)
df_burundi_factors_diff
```

**Improved version:** The `.tranpose()` function swaps a DataFrame's rows and columns. Row labels become column headers, and column headers become row labels. You can also use the shorthand `.T` property. After this step, the two columns exist but have generic names. We can overwrite this by assigning the desired column names using `.columns`. At the end, we can remove the '_increase' suffix in the score increase names (due to the direct tranposing) using the `removesuffix()` string method.

```python
df_burundi_factors_diff = max_diff_row[factor_increases_names].transpose().reset_index()
df_burundi_factors_diff.columns = ['factor', 'score_increase']
df_burundi_factors_diff['factor'] = df_burundi_factors_diff['factor'].str.removesuffix('_increase')

df_burundi_factors_diff
```
